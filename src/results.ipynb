```python
from Algorithm import Algorithm
import numpy as np
import DataUtil

Ks = np.arange(4, 10, 1)
length = 100

timing = []
result_dfs = []

for k in Ks:
    total_data_bits = k * 4e3
    D = DataUtil.generate_D_k_distribution(length, k) * total_data_bits
    D = D.numpy()
    H = DataUtil.generate_channel_gains(length, k)
    H = H.numpy()
    print('K =',k,'\n---------------')
    algo = Algorithm(D, H, None, None)
    algo.optimize('new')
    result_dfs.append(algo.to_df())
    timing.append(algo.get_time_used())
    break
```
